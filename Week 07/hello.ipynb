```python
# File Handling (Basic)

file_1 = open("sample_01.txt", "r")
print(file_1.read())
file_1.close()
```

```python
# used "with" (Modern)

with open("sample_01.txt", "r") as file_2: #Dont Close
    print(file_2.read())
```

```python
#Readline() method -> line by line read

with open("sample_01.txt", "r") as file:
    line_1 = file.readline()
    print(line_1)

    line_2 = file.readline()
    print(line_2)

    line_3 = file.readline()
    print(line_3)
```

```python
#.rstrip("\n") used

with open("sample_01.txt", "r") as file:
    line_1 = file.readline().rstrip("\n")
    print(line_1)

    line_2 = file.readline().rstrip("\n")
    print(line_2)

    line_3 = file.readline().rstrip("\n")
    print(line_3)
```

```python
with open("sample_01.txt", "r") as file:
    content = file.read()
    print(content, end="")
```

```python
# Second Type

with open("sample_01.txt", "r") as file:
    while True:
        line_1 = file.readline()

        if not line_1:
            break

        print(line_1.strip())
```

```python
with open("sample_01.txt", "r") as file:
    content = file.readline()
    print(content)# This will be a list of lines fron the file
    print(type(content)) # This will show that content is a list
```
