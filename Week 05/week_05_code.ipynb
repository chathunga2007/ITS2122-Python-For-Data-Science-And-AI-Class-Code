```python
# Function
def multiply(a, b):
    return a * b

print(multiply(4,5))
```

```python
# Positional Arguments
def student(name, age):
    print("Name:", name)
    print("Age: ", age)
student("Kamal", 56)
```

```python
# Default Arguments
def greet(name="Guest"):
    print("Hello!", name)

greet()
greet("Anil")
```

```python
def my(b, a=10):
    return a + b
print(my(1))

# def my(a=10, b): don't
```

```python
# Keyword Arguments
def fruits(name, price):
    print("Name: ",name)
    print("Price: ",price)

fruits(price=150, name="Apple")
```

```python
def my_fucntion(a, b, c):
    print("a: ",a)
    print("b: ",b)
    print("c: ",c)

my_fucntion(a = 1, b = 2, c = 3)
print("------------------------")
my_fucntion(b = 2, a = 1, c = 3)
print("------------------------")
my_fucntion(c = 3, b = 2, a = 1)
```

```python
# Arbitary Arguments
def my_fucntion(*args):
    print(args)
    print(type(args))

my_fucntion(2, 5, 90)
print("------------------------")
my_fucntion(10, 20, 30, 40, 50)
print("------------------------")
my_fucntion("Apple", "Orange", "PinaApple", "Cocomeloan")
```

```python
# Arbitary Keywords Argument
def my_function(**kwargs):
    print(kwargs)
    print(type(kwargs))

my_function(name="Chathunga", age=18, address="Galle")
```

```python
def my_function(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")

my_function(name="Alice", age=20, city="New York")
```

```python
# Absolute Value
print(abs(-20))
print(abs(20))
```

```python
# Map built in function
def square(x):
    return x * x

numbers = [1, 2, 3, 4, 5]
squared_number = list(map(square, numbers))
print(squared_number)
```

```python
# Map with Len
words = ["Hello", "World", "Python"]
words_length = list(map(len, words))
print(words_length)
```
