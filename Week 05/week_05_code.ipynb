```python
# Function
def multiply(a, b):
    return a * b

print(multiply(4,5))
```

```python
# Positional Arguments
def student(name, age):
    print("Name:", name)
    print("Age: ", age)
student("Kamal", 56)
```

```python
# Default Arguments
def greet(name="Guest"):
    print("Hello!", name)

greet()
greet("Anil")
```
