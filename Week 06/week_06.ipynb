```python
# Inharitance
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")

class Dog(Animal):
    def speak(self):
        print(f"{self.name} barks.")

# Example execution:
generic_animal = Animal("Creature")
dog = Dog("Buddy")

generic_animal.speak()  # Creature makes a sound.
dog.speak()             # Buddy barks.
```

```python
class Vehicle:
    def __init__(self):
        print("Super class constructor called!")
    def display_info(self):
        print("This is a super class method")

class Car(Vehicle):
    def __init__(self):
        print("Sub class constructor called!")
    def display_info(self):
        print("This is a sub class method")

# Creating Object
c1 = Car()
c1.display_info()
```

```python
# Base class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")


# Derived class
class Student(Person):
    def __init__(self, name, age, course):
        super().__init__(name, age)  # calling parent constructor
        self.course = course

    def show_student_info(self):
        self.show_info()


student1 = Student("Kamal", 20, "Computer Science")
student1.show_student_info()
```

```python
# Polymorphism example
class Shapes:
    def display_info(self):
        print("This is a shape.")

class Square(Shapes):
    def display_info(self):
        print("This is a square.")

class Circle(Shapes):
    def display_info(self):
        print("This is a circle.")

# Creating objects
Square_1 = Square()
Circle_1 = Circle()

# Calling methods
Square_1.display_info()  # Output: This is a square.
Circle_1.display_info()  # Output: This is a circle.
```

```python
# Encapsulation
# Public Class
class Employee:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def display_info(self):
        print("Name:", self.name)
        print("Age:", self.age)

employee1 = Employee("Namal", 50)
employee1.display_info()
```

```python
# Protected Class
class Employee:
    def __init__(self, name, age):
        self._name = name
        self._age = age
    def display_info(self):
        print("Name:", self._name)
        print("Age:", self._age)
    def _display_protected_info(self):
        print("This is a protected method")

employee1 = Employee("Namal", 50)
employee1.display_info()
employee1._display_protected_info() # Accessing protecting method(Not Reccomended)
employee1._name = "Nimal" # Accessing protecting attribute(Not Reccomended)
```

```python
# Private Class
class Employee:
    def __init__(self, name, age):
        self.__name = name # Private Attribute
        self.__age = age # Private Attribute
    def display_info(self):
        print("Name:", self.__name)
        print("Age:", self.__age)
    def __display_protected_info(self):
        print("This is a protected method")

employee1 = Employee("Namal", 50)
employee1.display_info()
# print(employee1.__name) # Accessing Private attribute(will raise and error)
```

```python
class Employee:
    def __init__(self, name, age):
        self.__name = name  # private attribute
        self.__age = age    # private attribute

    def display_info(self):
        print(f"Name: {self.__name}")
        print(f"Age: {self.__age}")

    # Getter method for name
    def get_name(self):
        return self.__name

    # Setter method for name
    def set_name(self, name):
        self.__name = name

    # Getter method for age
    def get_age(self):
        return self.__age

    # Setter method for age
    def set_age(self, age):
        self.__age = age


# Creating an object
emp1 = Employee("John Doe", 25)

# Display initial info
emp1.display_info()

# Accessing attributes using getters
print(emp1.get_name())
print(emp1.get_age())

emp1.set_name("Jane Doe")
emp1.set_age(30)

# Display updated info
emp1.display_info()
```
