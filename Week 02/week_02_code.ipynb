```python
tuple_1 = ("tennis",)
print(type(tuple_1))

tuple_2 = ("tennis")
print(type(tuple_2))
```

```python
# tuples with only one item
single_item_tuple = (10,)
print(type(single_item_tuple))

single_item_tuple = (10)
print(type(single_item_tuple))
```

```python
my_tuple = ("tennis", 10, True, "Vollyball", False, 80)
print(my_tuple[-1]) # negative indexing support tuples
```

```python
# range of the tuple
my_tuple = ("tennis", 10, True, "Vollyball", False, 80)
print(my_tuple[2:5])
print(my_tuple[:4])
print(my_tuple[2:])
```

```python
x = (1, 2, 3, 4, 5)

y = list(x)
print("Converted list:", y)
print("Type of converted list:", type(y))

y[2] = 10
print("Modified list:", y)

z = tuple(y)
print("Converted back to tuple:", z)

print("Type of converted tuple:", type(z))
```

```python
tuple_1 = (1, 2, 3, 4, 5, 6)
tuple_2 = list(tuple_1)
print(type(tuple_2))

tuple_2.remove(3)

tuple_1 = tuple(tuple_2)
print(tuple_1)
print(type(tuple_1))
```

```python
# Nested Tuple
nested_tuple = ((1, 2, 3), ("a", "b", "c"), (True, False))
print(nested_tuple)
print(nested_tuple[0])
print(nested_tuple[0][1])
print(nested_tuple[1][2])
print(nested_tuple[2][0])
```

```python
sports = {"tennis", "cricket", "football"}

# Adds a single item
sports.add("soccer")
print(sports)  # Output might be: {'tennis', 'cricket', 'football', 'soccer'}

# Adds multiple items
sports.update(["basketball", "volleyball"])
print(sports)  # Output might be: {'tennis', 'cricket', 'football', 'soccer', 'basketball', 'volleyball'}
```
