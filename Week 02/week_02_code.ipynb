```python
my_list = ["car", "bike", "bus", "train"]
my_list.pop(1) # remove the value from list
print(my_list) 
```

```python
current_list = ["cat"]
current_list.extend(["dog", "rabbit"])
print(current_list)
```

```python
my_list = ["car", "bike", "bus", "train"]
del my_list[1]
print(my_list)
```

```python
my_list = ["cat", "dog", "rat", "fish"]
my_list.clear() # list empty
print(my_list) 
```

```python
my_list = [10, 8, 2, 5, 7, 1]
my_list.sort() # accending order
print(my_list) 
```

```python
my_list = [10, 8, 2, 5, 7, 1]
my_list.sort(reverse=True) # decending order
print(my_list) 
```

```python
my_list = [10, 8, 2, 5, 7, 1]
my_list.reverse() # reverse the list
print(my_list)
```

```python
my_list = ["car", "bike", "bus", "train"]
new_list = my_list.copy()
print(new_list)
```

```python
list1 = ["car", "bike", "bus"]
list2 = ["train", "van", "lorry"]
joined_list = list1 + list2 # joined the list
print(joined_list)
```

```python
list1 = ["car", "bike", "bus"]
list2 = ["A", "B", "C"]
list1.extend(list2)
print(list1)
```

```python
sports = ("cricket", "football", "vollyball", "chess") #length of tuple
print(len(sports))
print(type(sports))
```

```python
tuple_1 = ("tennis",)
print(type(tuple_1))

tuple_2 = ("tennis")
print(type(tuple_2))
```

```python
# tuples with only one item
single_item_tuple = (10,)
print(type(single_item_tuple))

single_item_tuple = (10)
print(type(single_item_tuple))
```

```python
my_tuple = ("tennis", 10, True, "Vollyball", False, 80)
print(my_tuple[-1]) # negative indexing support tuples
```

```python
# range of the tuple
my_tuple = ("tennis", 10, True, "Vollyball", False, 80)
print(my_tuple[2:5])
print(my_tuple[:4])
print(my_tuple[2:])
```

```python
tuple_1 = (10, 20, 30)
tuple_2 = list(tuple_1) # convert to list

tuple_2.insert(2, 20.5)

tuple_1 = tuple(tuple_2) # convert to tuple

print(tuple_1)
print(type(tuple_1))
```
