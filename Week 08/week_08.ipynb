```python
# JSON File Load
import json

with open("employee.json", "r") as file_2:
    data = json.load(file_2)
    print(data)
    print(type(data))
```

```python
# JSON String Loads
import json

data = '{"name": "Kamal", "age": 18, "course": "Computer Science"}'

student = json.loads(data)

print(student)
print(student["name"])
```

```python
# Python Obj to JSON File with dump
import json

student = {
    "name": "Chathunga Bimsara",
    "age": 18,
    "course": "Software Engineering",
    "marks": 95
}

with open("student.json", "w") as file:
    json.dump(student, file)
```

```python
import json

student = {
    "name": "Chathunga Bimsara",
    "age": 18,
    "course": "Software Engineering",
    "marks": 95
}

data = json.dumps(student)

print(data)
print(type(data)) # This will show that data is a string
```

```python
# CSV -> List
import csv

with open("employees_01.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
# CSV -> Dictionary
import csv 

with open("employees_01.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
```

```python
import csv 

with open("employees_01.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["name"])
        print(row["department"])
        print(row["salary"])
```

```python
import csv

employee = [
    {
        "id": "E001",
        "name": "Shaluka",
        "department": "IT",
        "salary": 80000
    },
    {
        "id": "E002",
        "name": "Chathunga",
        "department": "IT",
        "salary": 150000
    }
]

with open("new_employees.csv", "w", newline="") as file:

    filedNames = ["id", "name", "department", "salary"]

    writer = csv.DictWriter(file, fieldnames=filedNames)

    writer.writeheader()

    writer.writerows(employee)
```
