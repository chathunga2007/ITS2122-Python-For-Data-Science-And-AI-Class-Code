```python
import csv

employee = [
    {
        "id": "E001",
        "name": "Shaluka",
        "department": "IT",
        "salary": 80000
    },
    {
        "id": "E002",
        "name": "Chathunga",
        "department": "IT",
        "salary": 150000
    }
]

with open("new_employees.csv", "w", newline="") as file:

    filedNames = ["id", "name", "department", "salary"]

    writer = csv.DictWriter(file, fieldnames=filedNames)

    writer.writeheader()

    writer.writerows(employee)
```

```python
# Numpy Library
# Create Array

import numpy as np
a = np.array([1, 2, 3, 4, 5])
print("Array:",a)
```

```python
#Basic Operations

a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

print("Addition:", a + b)
print("Substraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
# Square root, mean, max & min
a = np.array([1, 4, 8, 11, 25, 50])

print("Square root:", np.sqrt(a))
print("Mean:", np.mean(a))
print("Max:", np.max(a))
print("Min:", np.min(a))
```

```python
# 2D Array
a = np.array([[1, 2, 3], [4, 5, 6]])

print("2D Array:\n", a)
print("Shape:", a.shape)
print("Size:", a.size)
```

```python
# 3D Array
a = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("3D Array:\n", a)
print("Shape:", a.shape)
print("dtType:", a.dtype)
```

```python
# Array indexing 
# 1D Array Indexing

a = np.array([10, 20, 30, 40, 50])
print("Element at index 2:", a[2])
```

```python
# 3D array indexing

a = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("3D Array:\n", a)
print("Element at depth 1, row 0, column 1:", a[1, 0, 1])
```

```python
# Negative Indexing
a = np.array([10, 20, 30, 40, 50])
print("Last Element:", a[-1])

b = np.array([[1, 2, 3], [4, 5, 6]])
print("Last Row:", b[-1])
print("Last Column:", b[:, -1])
```

```python
# Boolean Indexing

a = np.array([10, 20, 30, 40, 50])

marks = a > 25

print("Element greater than 25:", a[marks])
```

```python
# Multiple Conditions

a = np.array([10, 20, 30, 40, 50])

condition = (a > 20) & (a < 50)

print("Elements between 20 and 50:", a[condition])
```

```python
# Array Inialization 2D Array
a = np.zeros((2, 3))
print("Zeors Array:\n", a)

a = np.ones((2, 3))
print("Ones Array:\n", a)
```

```python
a = np.zeros((2, 3), dtype=int)
print("Zeros array with float type:\n", a)
```

```python
# numpy.full
a = np.full((2, 3), 7)
print("Full Array:\n", a)
```

```python
# Create Dataframe
import pandas as pd
data = {
    "Name": ["Tharusha", "Shaluka", "Naveen", "Chathunga"],
    "Age": [20, 23, 22, 18],
    "City": ["New York", "Galle", "Colombo", "Australia"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Displaying the first few rows of the DataFrame
print("\nFirst 2 Rows:")
print(df.head(2))
```

```python
# Access cells using .loc & .iloc
# Accessing a single cell using .loc
print("\nAccessing a single cell using .loc:")
print(df.loc[1, "Name"])

# Accessing a single cell using .iloc
print("\nAccessing a single cell using .iloc:")
print(df.iloc[1, 0])
```

```python
# Named Index
df.index = ["Row01", "Row02", "Row03", "Row04"]
print("\nDataFrame with Named Index:")
print(df)
```

```python
# Read Column
print("\nColumn Names:")
print(df.columns)
print(list(df))
```

```python
# Get Values
print("\nValues in DataFrame:")
print(df.values)
```

```python
# Get Data Frame Rows
print("\nShape of DataFrame:")
print(df.shape)
```

```python
# Get Data Types
print("\nData Types of Each Column:")
print(df.dtypes)
```

```python
# Get Element Count
print("\nTotal Number of Elements in DataFrame:")
print(df.size)
```

```python
# Get Length
print("\nNumber of Rows in DataFrame:")
print(len(df))
```

```python
# Get Sum
print("\nSum of Age:")
print(df["Age"].sum())

print(df["Age"].mean())
```

```python
# Get Info
print("\nDataFrame Info:")
df.info()
```

```python
# Get Description
print("\nDescription Statistics:")
print(df.describe())
```

```python
# Filtering DataFrame
# Filter rows where Age is greater than 30

filtered_df = df[df["Age"] > 30]
print("\nFiltered DataFrame (Age > 30):")
print(filtered_df)
```
