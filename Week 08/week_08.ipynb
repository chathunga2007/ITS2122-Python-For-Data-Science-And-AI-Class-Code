```python
# JSON File Load
import json

with open("employee.json", "r") as file_2:
    data = json.load(file_2)
    print(data)
    print(type(data))
```

```python
# JSON String Loads
import json

data = '{"name": "Kamal", "age": 18, "course": "Computer Science"}'

student = json.loads(data)

print(student)
print(student["name"])
```

```python
# Python Obj to JSON File with dump
import json

student = {
    "name": "Chathunga Bimsara",
    "age": 18,
    "course": "Software Engineering",
    "marks": 95
}

with open("student.json", "w") as file:
    json.dump(student, file)
```

```python
import json

student = {
    "name": "Chathunga Bimsara",
    "age": 18,
    "course": "Software Engineering",
    "marks": 95
}

data = json.dumps(student)

print(data)
print(type(data)) # This will show that data is a string
```
