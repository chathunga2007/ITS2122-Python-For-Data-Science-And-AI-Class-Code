```python
# Python Obj to JSON File with dump
import json

student = {
    "name": "Chathunga Bimsara",
    "age": 18,
    "course": "Software Engineering",
    "marks": 95
}

with open("student.json", "w") as file:
    json.dump(student, file)
```

```python
import csv

employee = [
    {
        "id": "E001",
        "name": "Shaluka",
        "department": "IT",
        "salary": 80000
    },
    {
        "id": "E002",
        "name": "Chathunga",
        "department": "IT",
        "salary": 150000
    }
]

with open("new_employees.csv", "w", newline="") as file:

    filedNames = ["id", "name", "department", "salary"]

    writer = csv.DictWriter(file, fieldnames=filedNames)

    writer.writeheader()

    writer.writerows(employee)
```

```python
# Numpy Library
# Create Array

import numpy as np
a = np.array([1, 2, 3, 4, 5])
print("Array:",a)
```

```python
#Basic Operations

a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

print("Addition:", a + b)
print("Substraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
# Square root, mean, max & min
a = np.array([1, 4, 8, 11, 25, 50])

print("Square root:", np.sqrt(a))
print("Mean:", np.mean(a))
print("Max:", np.max(a))
print("Min:", np.min(a))
```

```python
# 2D Array
a = np.array([[1, 2, 3], [4, 5, 6]])

print("2D Array:\n", a)
print("Shape:", a.shape)
print("Size:", a.size)
```

```python
# 3D Array
a = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("3D Array:\n", a)
print("Shape:", a.shape)
print("dtType:", a.dtype)
```

```python
# Array indexing 
# 1D Array Indexing

a = np.array([10, 20, 30, 40, 50])
print("Element at index 2:", a[2])
```

```python
# 3D array indexing

a = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("3D Array:\n", a)
print("Element at depth 1, row 0, column 1:", a[1, 0, 1])
```

```python
# Negative Indexing
a = np.array([10, 20, 30, 40, 50])
print("Last Element:", a[-1])

b = np.array([[1, 2, 3], [4, 5, 6]])
print("Last Row:", b[-1])
print("Last Column:", b[:, -1])
```

```python
# Boolean Indexing

a = np.array([10, 20, 30, 40, 50])

marks = a > 25

print("Element greater than 25:", a[marks])
```

```python
# Multiple Conditions

a = np.array([10, 20, 30, 40, 50])

condition = (a > 20) & (a < 50)

print("Elements between 20 and 50:", a[condition])
```

```python
# Array Inialization 2D Array
a = np.zeros((2, 3))
print("Zeors Array:\n", a)

a = np.ones((2, 3))
print("Ones Array:\n", a)
```

```python
a = np.zeros((2, 3), dtype=int)
print("Zeros array with float type:\n", a)
```

```python
# numpy.full
a = np.full((2, 3), 7)
print("Full Array:\n", a)
```
