```python
# IF Statement
x = 10
if x > 5:
    print("x is greater than 5")
```

```python
# IF-ELSE Statement
x = 3
if x > 5:
    print("Greater")
else:
    print("Not greater")
```

```python
# IF-ELSE-IF Statement
num = int(input("Enter a number: "))

if num > 0:
    print("Positive number")
elif num == 0:
    print("Zero")
else:
    print("Negative number")
```

```python
# Nested IF Statement
age = 18
has_id = True

if age >= 18:
    if has_id:
        print("You can enter")
    else:
        print("ID required")
else:
    print("You are underage")
```

```python
# Ternary Operator
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)
```

```python
# Structural Pattern Matching
response_code = 502
match response_code:
    case 200 | 201:
        print("OK")
    case 300 | 307:
        print("Redirect")
    case 400 | 401:
        print("Bad Request")
    case 500 | 502:
        print("Internal Server Error")
```

```python
my_list = [1, 2, 3]
match my_list:
    case []:
        print("The list is empty")
    case [x]:
        print("The list has one element: ", x)
    case [x, y]:
        print("The list has two elements: ", x, y)
    case _:
        print("The list has more than two elements")
```

```python
# For Loop
my_list = ["Apple", "Banana", "Cherry"]
for item in my_list:
    print(item)
```

```python
num = [x for x in range(5)]
print(num)
```

```python
even_num = [x for x in range(11) if x % 2 == 0]
print(even_num)
```

```python
squares_number = [x**2 for x in range(5)]
print(squares_number)
```

```python
sequence_num = [x ** 3 for x in range(1, 21) if x % 3 == 0]
print(sequence_num)
```

```python
# While Loop
i = 1
while i <= 5:
    print(i)
    i+=1
```

```python
my_list = {
    "Name": "Chathunga Bimsara",
    "City": "Galle",
    "Age": 18 
}

for key in my_list.keys():
    print("Key:", key)

print("-------------------")

for value in my_list.values():
    print("Value:", value)

print("-------------------")

for key, value in my_list.items():
    print("Key:", key, "Value:", value)
```

```python
# Break Statement
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

```python
# Continue Statement
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
# Loop with Else
for i in range(1, 5):
    if i == 3:
        break
    print(i)
else:
    print("Loop Completed Successfully")
```
