```python
# IF Statement
x = 10
if x > 5:
    print("x is greater than 5")
```

```python
# IF-ELSE Statement
x = 3
if x > 5:
    print("Greater")
else:
    print("Not greater")
```

```python
# IF-ELSE-IF Statement
num = int(input("Enter a number: "))

if num > 0:
    print("Positive number")
elif num == 0:
    print("Zero")
else:
    print("Negative number")
```

```python
# Nested IF Statement
age = 18
has_id = True

if age >= 18:
    if has_id:
        print("You can enter")
    else:
        print("ID required")
else:
    print("You are underage")
```

```python
# Ternary Operator
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)
```
