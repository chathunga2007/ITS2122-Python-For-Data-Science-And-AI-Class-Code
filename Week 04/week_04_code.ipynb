```python
# IF Statement
x = 10
if x > 5:
    print("x is greater than 5")
```

```python
# IF-ELSE Statement
x = 3
if x > 5:
    print("Greater")
else:
    print("Not greater")
```

```python
# IF-ELSE-IF Statement
num = int(input("Enter a number: "))

if num > 0:
    print("Positive number")
elif num == 0:
    print("Zero")
else:
    print("Negative number")
```

```python
# Nested IF Statement
age = 18
has_id = True

if age >= 18:
    if has_id:
        print("You can enter")
    else:
        print("ID required")
else:
    print("You are underage")
```

```python
# Ternary Operator
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)
```

```python
# Structural Pattern Matching
response_code = 502
match response_code:
    case 200 | 201:
        print("OK")
    case 300 | 307:
        print("Redirect")
    case 400 | 401:
        print("Bad Request")
    case 500 | 502:
        print("Internal Server Error")
```

```python
my_list = [1, 2, 3]
match my_list:
    case []:
        print("The list is empty")
    case [x]:
        print("The list has one element: ", x)
    case [x, y]:
        print("The list has two elements: ", x, y)
    case _:
        print("The list has more than two elements")
```

```python
# For Loop
my_list = ["Apple", "Banana", "Cherry"]
for item in my_list:
    print(item)
```

```python
num = [x for x in range(5)]
print(num)
```

```python
even_num = [x for x in range(11) if x % 2 == 0]
print(even_num)
```

```python
squares_number = [x**2 for x in range(5)]
print(squar_number)
```
