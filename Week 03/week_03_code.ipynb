```python
# Arithmetic Operators
a = 10
b = 20

print(a + b)
print(b - a)
print(a * b)
print(b / a)
print(a // b)
print(a % b)
print(a ** b)
```

```python
# Assignment Operators
x = 5
print(x)

x = 5
x*=2
print(x)

x = 5
x/=2
print(x)
```

```python
# Logical Operators
x, y = 5, 10
print(x > 1 and y < 10)

x, y = 5, 10
print(x > 6 or y < 10)

x = 5
print(not x < 1)
```

```python
# Identity Operators
x = 5
y = 50

print(x is y)
print(x is not y)
```

```python
# Membership Operators
num = [1, 2, 3, 4, 5]
print(1 in num)
print(10 in num)

num = [1, 2, 3, 4, 5]
print(1 not in num)
print(10 not in num)
```

```python
# Bitwise Operators
a = 10  # Binary: 1010
b = 4   # Binary: 0100

print(a & b)  # 0000 -> Output: 0
print(a | b)  # 1110 -> Output: 14
print(a ^ b)  # 1110 -> Output: 14
print(~10)   # -(10 + 1) -> Output: -11
print(~-10)  # -(-10 + 1) -> Output: 9
```

```python
# Operators Precedence 
result = not 10 + 5 <= 15 or 20 / 4 == 5 and 2 ** 3 > 7
print(result)
```
