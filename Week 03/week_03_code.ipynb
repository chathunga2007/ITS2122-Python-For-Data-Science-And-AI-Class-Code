```python
# Arithmetic Operators
a = 10
b = 20

print(a + b)
print(b - a)
print(a * b)
print(b / a)
print(a // b)
print(a % b)
print(a ** b)
```

```python
# Assignment Operators
x = 5
print(x)

x = 5
x*=2
print(x)

x = 5
x/=2
print(x)
```

```python
# Logical Operators
x, y = 5, 10
print(x > 1 and y < 10)

x, y = 5, 10
print(x > 6 or y < 10)

x = 5
print(not x < 1)
```
